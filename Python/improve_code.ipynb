Write a program that divides two numbers and handles division by zero.


```python
def divied(a,b):
    try: 
        if b>0:
            print(f'{a}/{b}=',a/b)
        else:
            raise ZeroDivisionError
    except ZeroDivisionError:
        print('you cant divied by zero')
divied(5,4)
```

## Decorater

```python
def my_decorater(function):
    def wrapper():
        print('before function')
        function()
        print('after function')
    return wrapper
@my_decorater
def greet():
    print('Hello')

greet()
```

```python
def my_decorater(function):
    def wrapper(*arg,**kwarg):
        print('before function')
        result=function(*arg,**kwarg)
        print('after function')
        return result
    return wrapper

@my_decorater
def add(a,b):
    print(f'the sum of {a} and {b} is', a+b)

add(15,10)
```

## Generator

```python
def count_number(limit):
    number=0
    while number<=limit:
        yield number

        number+=1
p=count_number(11)
for i in range(0,10):
    print(next(p))
```

1. Create a decorator that prints "Function started" before a function runs.


```python
def decorator(function):
    def wrapper():
        print('function started')
        function()
    return wrapper

@decorator
def greet():
    print(" hello \n i am ajay \n AI engineer")

greet()
```

2. Create a decorator that prints "Function finished" after a function runs.


```python
def decorator(func):
    def wrapper(a,b):
        result=func(a,b)
        print(result)
        print('function finished')
    return wrapper
@decorator
def sum(a,b):
    return f'sum of {a} and {b} is {a+b}'

sum(11,22)
```

3. Write a decorator for a greet() function that prints a welcome message before "Hello".


```python
def welcome_msg(func):
    def wrapper(a):
        print('welcome to  my coding world')
        func(a)
    return wrapper
@welcome_msg
def greet(a):
    print(f'Hello i am .{a}')

greet('Raj')
```
