1. NumPy: Create a NumPy array of 20 random integers (1–100). Without using any loops, print all
values above the mean, and multiply the whole array by 2 using vectorization. Explain in a
comment what broadcasting did.

```python
import numpy as np
import random
array=np.random.randint(1,100,size=20)
b=array[array>np.mean(array)]
print(b)
print(b*2)
```

2. Pandas basics: Build a DataFrame of 6 employees with name, department, and salary. Filter
employees earning above 50,000, then use groupby to show the average salary per department.


```python
import pandas as pd
df=pd.DataFrame({
    'name':['ram','karn','sham','arjun','akshay','veer'],
    'department':['math','phy','math','math','bio','bio'],
    'salary':[55000,40000,22000,35000,57000,59000]

})
high_salary=df[df['salary']>50000]
print(high_salary)
avg_salary=df.groupby('department')['salary'].mean()
print(avg_salary)
```

3. Cleaning & merging: Given two DataFrames — one with student IDs + names, one with student
IDs + marks (containing some NaN values) — merge them, fill missing marks with the column
mean, and drop any duplicate rows.

```python
import numpy as np
d1=pd.DataFrame({
    'std_id':[1,2,3,4,5],
    'name' :['ram','karn','sham','arjun','akshay']

})
d2=pd.DataFrame({
    'std_id':[1,2,3,4,5,3],
    'marks' :[56,np.nan,78,np.nan,66,78]

})
new_data=pd.merge(d1,d2, on='std_id', how='inner')
print(new_data)
marks_mean=new_data['marks'].mean()
new_data['marks']=new_data['marks'].fillna(marks_mean)
new_data=new_data.drop_duplicates()
print(new_data)
```
