1.OOP: Create a BankAccount class with a private balance, deposit/withdraw methods with validation (no negative amounts, no overdraft), and a SavingsAccount subclass that adds interest. Demonstrate encapsulation and inheritance with a short usage example.


```python
class BankAccount():
    def __init__(self,balance=5000):
        self.__balance=balance
    def get_balance(self):
        return self.__balance
    def deposit(self,amount):
        self.__balance+= amount
        return f'you deposit amount is {amount} and your current balance is {self.__balance}'
    def withdraw(self,amount):
        if self.__balance>amount:
            self.__balance-=amount
            return f'you withdraw amount is {amount} and your current balance is {self.__balance}'
        else:
            return 'low balance'
class SavingsAccount(BankAccount):
    @BankAccount
    def accinfo():
        return f'current balance is'
p1=SavingsAccount()
p1.get_balance()
p1.withdraw(200)
#p1.deposit(200)
```

```python
import time
def time_func(func):
    def wrapper():
        start=time.time()
        result=func()
        end=time.time()
        print('exact time is ', end-start)
        return result
    return wrapper


list=[0,1]
@time_func
def fibonacci_num(limit=20):
    for i in range(limit-2):
        list.append(list[-1]+list[-2])
    print(list)
fibonacci_num()
```
